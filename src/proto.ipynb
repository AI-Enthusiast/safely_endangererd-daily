```python
url = 'https://safelyendangered.com/'
```

```python
raw = requests.get(url)
if raw.status_code == 200:
    print('Request successful')
else:
    print('Request failed with status code:', raw.status_code)
```

```python
soup = bs(raw.text, 'html.parser')
```

```python
# get the main comic image page
# in this case it is <a href="/blogs/comics/snowball-fight" class="full-unstyled-link">
#               Snowball Fight
#             </a>
comic_li = soup.find('li', id=lambda x: x and x.startswith('Slide-template--17273659850913__featured_blog_'))
comic_link = comic_li.find('a', class_='full-unstyled-link')
comic_url = 'https://safelyendangered.com' + comic_link['href']
comic_title = comic_link.text.strip()
print('Comic URL:', comic_url)
print('Comic Title:', comic_title)
```

```python
comic_page = requests.get(comic_url)
if comic_page.status_code == 200:
    print('Comic page request successful')
else:
    print('Comic page request failed with status code:', comic_page.status_code)
```
